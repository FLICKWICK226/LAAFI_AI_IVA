# 📓 Notebook 01 : Setup, Vérification Environnement & Secrets Colab
**Projet :** LAAFI_AI IVA Engine (Version 2.0)
**Objectif :** Valider l'accès GPU, monter le Drive, charger les secrets Kaggle et fixer la graine SEED = 42.

```python
# 1. Montage du Drive et Configuration du Répertoire de Travail
import os
import sys

if 'google.colab' in sys.modules:
    from google.colab import drive
    drive.mount('/content/drive')
    project_path = '/content/drive/MyDrive/LAAFI_AI_IVA'
    os.makedirs(project_path, exist_ok=True)
    os.chdir(project_path)
    sys.path.append(project_path)
    print(f"✅ Répertoire de travail Colab : {os.getcwd()}")
else:
    print(f"💻 Environnement Local détecté : {os.getcwd()}")
```

```python
# 2. Fixation de la Graine Globale SEED = 42
try:
    from src.utils.seed import seed_everything
    seed_everything(42)
except ImportError:
    import random, numpy as np, torch
    random.seed(42)
    np.random.seed(42)
    torch.manual_seed(42)
    print("🔒 Graine fixée manuellement à 42")
```

```python
# 3. Vérification GPU & VRAM
import torch

print(f"PyTorch Version : {torch.__version__}")
if torch.cuda.is_available():
    device_name = torch.cuda.get_device_name(0)
    vram_gb = torch.cuda.get_device_properties(0).total_memory / 1e9
    print(f"🚀 GPU Détecté : {device_name} ({vram_gb:.2f} GB VRAM)")
else:
    print("⚠️ Aucun GPU détecté. L'entraînement nécessitera un GPU Tesla T4 sur Colab.")
```

```python
# 4. Récupération des Secrets Kaggle via Colab Userdata
if 'google.colab' in sys.modules:
    try:
        from google.colab import userdata
        os.environ["KAGGLE_USERNAME"] = userdata.get("KAGGLE_USERNAME")
        os.environ["KAGGLE_KEY"] = userdata.get("KAGGLE_KEY")
        print("🔑 Clés API Kaggle chargées avec succès depuis userdata !")
    except Exception as e:
        print(f"❌ Erreur lors de la récupération des secrets Kaggle : {e}")
else:
    print("ℹ️ Mode Local : Assurez-vous d'avoir fixé KAGGLE_USERNAME et KAGGLE_KEY dans votre environnement.")
```
